```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/cleaned/financial_fraud_clean.csv")

df["Transaction_Date"] = pd.to_datetime(
    df["Transaction_Date"]
)

df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df["Transaction_Hour"] = df["Transaction_Date"].dt.hour
```

```python
df[["Transaction_Date", "Transaction_Hour"]].head()
```

```python
df["Transaction_Month"] = df["Transaction_Date"].dt.month
```

```python
df[["Transaction_Date", "Transaction_Month"]].head()
```

```python
## Amount Compared With Average Spend
```

```python
df["Amount_vs_Average"] = (
    df["Transaction_Amount"] / df["Average_Spend"]
)
```

```python
df[[
        "Transaction_Amount",
        "Average_Spend",
        "Amount_vs_Average"
    ]].head()
```

```python
df[["Transaction_Date","Transaction_Hour","Transaction_Month","Transaction_Amount","Average_Spend","Amount_vs_Average"]].head()
```

```python
categorical_columns = [
    "Merchant_Category",
    "Payment_Method",
    "Device_Type",
    "Location",
    "Suspicious_Keyword"
]

df[categorical_columns].head()
```

```python
for column in categorical_columns:
    print("\n", column)
    print(df[column].unique())
```

```python
## onehot encoding - it convert category column inti number 
```

```python
df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)
```

```python
df.head()
```

```python
df.columns
```

```python
## let create target var for feature Fraudulent is the value we want the ML models to predict.
```

```python
target = "Fraudulent"

X = df.drop(columns=[target])
y = df[target]
```

```python
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print(y.value_counts())
```

```python
X.dtypes
```

```python
X.isnull().sum().sum()
```

```python
print("Number of features:", X.shape[1])
print("\nFeature columns:")
print(X.columns.tolist())
```

```python
X = X.drop(
    columns=[
        "Transaction_ID",
        "Customer_ID",
        "Transaction_Date"
    ]
)
```

```python
print("Features shape:", X.shape)
print("\nFeature columns:")
print(X.columns.tolist())
```

```python
X.dtypes
```

```python
print("Missing values:", X.isnull().sum().sum())
print("Number of features:", X.shape[1])
print("Number of records:", X.shape[0])
```

```python
fraud_features = X.copy()

fraud_features["Fraudulent"] = y.values

fraud_features.head()
```

```python
print("Shape:", fraud_features.shape)
print("Missing values:", fraud_features.isnull().sum().sum())
print("Number of features:", fraud_features.shape[1] - 1)
print("Number of records:", fraud_features.shape[0])
```

```python
fraud_features.to_csv(
    "../data/cleaned/fraud_features.csv",
    index=False
)
```
